**3. Creating a BeautifulSoup Object**

Example

```python
import requests
a = requests.get('https://books.toscrape.com/')
```

```python
with open(f"htmls/page1.html","w") as f:
  f. write(a.text)
```

```python
from bs4 import BeautifulSoup
```

```python
with open("htmls/page1.html") as f:
  content = f.read()
```

```python
soup = BeautifulSoup(content,"lxml")
```

```python
h3=soup.find_all("h3")
h3
```

```python
for h3_tag in h3:
  print(h3_tag.find("a").get("title"))
```

```python
articles = soup.find_all("article",class_="product_pod")
articles
```

```python
itmes = []
for article in articles:

  title =article.find("h3").find("a")["title"]
  price = article.select_one("p.price_color").text.split("£")[1]
  rating_element = article.select_one("p.star-rating")
  rating = rating_element["class"][1]
  itmes.append([title,price,rating])
```

```python
itmes
```

```python
import pandas as pd
df =pd.DataFrame(itmes,columns=["Title","Price","rating"])
df
```

```python
df.to_csv("books.csv",index=False)

from google.colab import files
files.download("books.csv")
```
